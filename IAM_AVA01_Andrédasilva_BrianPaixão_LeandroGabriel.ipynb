O conjunto de dados apresenta 13 atributos preditivos relacionados às características físico-químicas dos vinhos. Esses atributos são representados por valores numéricos e, portanto, são classificados como **dados quantitativos**.

Quanto à escala de mensuração, os atributos quantitativos são considerados de **escala de razão**, pois representam medidas numéricas nas quais existe um zero significativo e são possíveis comparações de magnitude.

O atributo `target`, por sua vez, representa a classe do vinho. Embora esteja armazenado no Python como valores inteiros (`0`, `1` e `2`), esses valores são códigos que representam categorias (`class_0`, `class_1` e `class_2`). Portanto, o `target` é um dado **qualitativo/categórico**, de **escala nominal**.

```python
from sklearn.preprocessing import StandardScaler

X = df.drop(columns=["target"])
y = df["target"]

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(X_scaled, columns=X.columns)

df_scaled.head()
```

### Questão H) Transformação dos dados

Foi realizada a padronização dos atributos preditivos utilizando o método StandardScaler. Essa transformação foi escolhida porque os atributos do conjunto de dados apresentam escalas e amplitudes diferentes.

A padronização transforma os atributos para uma escala comparável, fazendo com que cada variável apresente média aproximadamente igual a zero e desvio-padrão aproximadamente igual a um.

Essa transformação é especialmente importante para técnicas que são influenciadas pela escala dos atributos, como a Análise de Componentes Principais (PCA), utilizada na questão seguinte.

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

X_pca.shape
```

```python
df_pca = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

df_pca["target"] = y.values

df_pca.head()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df_pca,
    x="PC1",
    y="PC2",
    hue="target",
    palette="Set1"
)

plt.title("PCA - Wine Dataset")
plt.xlabel("Componente Principal 1 (PC1)")
plt.ylabel("Componente Principal 2 (PC2)")

plt.show()
```

```python
pca.explained_variance_ratio_
```

### Questão I) Análise de Componentes Principais (PCA)

Foi aplicada a Análise de Componentes Principais (PCA) sobre os atributos preditivos previamente padronizados. Foram selecionadas duas componentes principais, PC1 e PC2, com o objetivo de representar os dados em um espaço bidimensional e facilitar sua visualização.

A primeira componente principal (PC1) apresentou uma variância explicada de aproximadamente 36,20%, enquanto a segunda componente (PC2) apresentou aproximadamente 19,21%. Dessa forma, as duas componentes juntas explicam aproximadamente 55,41% da variabilidade total presente nos dados.

O gráfico da PCA permite representar as 178 amostras em um espaço formado pelas componentes PC1 e PC2, possibilitando observar a distribuição das amostras e verificar visualmente a separação entre as diferentes classes de vinho.

Embora a redução de 13 para 2 dimensões implique perda de parte da informação, as duas componentes preservam aproximadamente 55,41% da variabilidade dos dados, proporcionando uma representação bidimensional útil para análise e visualização do conjunto.
